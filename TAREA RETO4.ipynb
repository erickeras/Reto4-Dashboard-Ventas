```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar los datos, el csv debe estar junto al jupyter notebook
df = pd.read_csv('supermarket_sales.csv')

# Limpieza básica, se convierte la columna Date a formato fecha para un mejor analisis 
df['Date'] = pd.to_datetime(df['Date'])
# Se elimian los valores nulos con el fin de evitar errores en las gráficas
df = df.dropna()

print("Datos listos y limpios. Total de filas:", df.shape[0])
```

### Preguntas de Negocio y Justificación de Gráficos

1. ¿Qué línea de producto tiene los mejores resultados?
* Gráfico seleccionado: Gráfico de Barras (Barplot).
* Información que se muestra en el gráfico: Los ingresos totales acumulados por cada línea de producto.
* Por qué fue seleccionado y cuándo es útil: Este gráfico da una visualización más adecuada para comparar volúmenes entre distintas categorías nominales. Esto resulta util para entornos de negocio en donde se debe identificar rápidamente que áreas generan más ingresos y cuales necesitan un impulso en ventas por ejemplo. 

2. ¿Cómo se distribuyen los ingresos por segmento?
* Gráfico seleccionado: Gráfico de Caja (Boxplot).
* Información que se muestra en el gráfico: Se muestra la distribución del gasto total, la mediana y los posibles valores atípicos, esto estará segmentado por el género del comprador. 
* Por qué fue seleccionado y cuándo es útil: Este gráfico se eligió porque permite visualizar la información más allá de solo el promedio, mostrando la dispersión real de las ventas, esto puede ser útil cuando se requiere entender el comportamiento de compra y detectar anomalías en los pagos de los clientes.

```python
# Se aplica el estilo visual global de Seaborn
sns.set_theme(style="whitegrid", palette="muted")

# Se crea el lienzo principal del Dashboard con 3 filas y 2 columnas
fig, axes = plt.subplots(3, 2, figsize=(16, 18))
fig.suptitle('Dashboard de Ventas del Supermercado', fontsize=20, fontweight='bold', y=0.98)


# 1. Gráfico de barras: Ingresos por Línea de Producto (Fila 0, Columna 0)
sns.barplot(data=df, x='Total', y='Product line', ax=axes[0, 0], errorbar=None, palette='viridis', hue='Product line', legend=False)
axes[0, 0].set_title('Ingresos Totales por Línea de Producto', fontsize=14)
axes[0, 0].set_xlabel('Ingreso Total ($)')
axes[0, 0].set_ylabel('Línea de Producto')

# 2. Gráfico de caja: Distribución de Gasto por Género (Fila 0, Columna 1)
sns.boxplot(data=df, x='Gender', y='Total', ax=axes[0, 1], palette='Set2', hue='Gender', legend=False)
axes[0, 1].set_title('Distribución de Gasto por Género', fontsize=14)
axes[0, 1].set_xlabel('Género')
axes[0, 1].set_ylabel('Gasto Total ($)')

plt.tight_layout()
plt.show()
```
